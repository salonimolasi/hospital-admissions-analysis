```python
# Step 1: calculate the average length of stay for each department
avg_los = df.groupby("department")["length_of_stay"].mean().sort_values()

# Step 2: create the chart
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(avg_los.index, avg_los.values, color="#2A9D8F")

# Step 3: add labels
ax.set_title("Average Length of Stay by Department")
ax.set_xlabel("Department")
ax.set_ylabel("Average stay (days)")
#legend
plt.grid(color="blue",linestyle="--")
fig.savefig('avg_los_by_department.png',dpi=300,bbox_inches="tight")
# Step 4: show the chart
plt.show()
```
